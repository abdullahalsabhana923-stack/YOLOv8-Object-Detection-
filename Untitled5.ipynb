```python
# 1. Setup and Installation
!pip install ultralytics

import os
import glob
from ultralytics import YOLO
from IPython.display import Image, display

# 2. Model Training (Object Detection with YOLOv8)
model = YOLO('yolov8n.pt')
results = model.train(data='coco8.yaml', epochs=10, imgsz=640)

# 3. Evaluation Metrics Visualization
print("\n=== Section 5: Evaluation Metrics ===")
results_path = 'runs/detect/train/results.png'
if os.path.exists(results_path):
    display(Image(filename=results_path))

# 4. Results: Prediction Cases
print("\n=== Successful and Failure Detections ===")
val_images = glob.glob('runs/detect/train/val_batch0_pred.jpg')
if val_images:
    display(Image(filename=val_images[0]))
```
